```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 2*np.pi, 200)
x = np.sin(t)

plt.plot(t, x)
plt.xlabel("Tiempo")
plt.ylabel("Posición")
plt.title("Movimiento armónico simple")
plt.grid(True)
plt.show()
```
